```python
import pandas as pd, matplotlib.pyplot as plt, numpy, torch
print(pandas.__version__, torch.__version__)
```

```python
url = "https://media.githubusercontent.com/media/MuseumofModernArt/collection/refs/heads/main/Artworks.csv"
df = pandas.read_csv(url)
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df['Diameter (cm)'].hist(bins=40)
plt.xlabel('Weight'); plt.ylabel('kg'); plt.title('Track Weight Distribution')
```

```python
plt.scatter(df['Length (cm)'], df['Height (cm)'], alpha=0.3)
plt.xlabel('Weight'); plt.ylabel('kg'); plt.title('Track Weight Distribution')
```

```python
decade = (pd.to_numeric(df['DateAcquired'].str[:4], errors='coerce') // 10 )
decade.value_counts().sort_index().plot(kind='bar')
```
